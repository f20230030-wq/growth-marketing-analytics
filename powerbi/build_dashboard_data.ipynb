```python
import os, pandas as pd
os.chdir(os.path.expanduser("~/growth-marketing-analytics"))

files = {
    # Module A: Criteo
    "A_overall_kpis":        "module_1_criteo/outputs/overall_kpis.csv",
    "A_campaigns":           "module_1_criteo/outputs/campaign_perf_stable.csv",
    "A_frequency":           "module_1_criteo/outputs/frequency_curve.csv",
    "A_attribution":         "module_1_criteo/outputs/attribution_by_campaign.csv",
    # Module B: Google Analytics
    "B_overall_kpis":        "module_2_google_analytics/outputs/overall_kpis.csv",
    "B_channels":            "module_2_google_analytics/outputs/channel_scorecard.csv",
    "B_funnel":              "module_2_google_analytics/outputs/funnel_overall.csv",
    "B_funnel_by_channel":   "module_2_google_analytics/outputs/funnel_by_channel.csv",
    "B_devices":             "module_2_google_analytics/outputs/device_performance.csv",
    "B_new_vs_returning":    "module_2_google_analytics/outputs/new_vs_returning.csv",
    "B_monthly":             "module_2_google_analytics/outputs/monthly_trend.csv",
    "B_campaigns":           "module_2_google_analytics/outputs/campaign_performance.csv",
    # Module C: Olist
    "C_origins":             "module_3_olist/outputs/origin_scorecard.csv",
    "C_activation_fair":     "module_3_olist/outputs/activation_fair.csv",
    "C_lead_types":          "module_3_olist/outputs/by_lead_type.csv",
    "C_rfm_segments":        "module_3_olist/outputs/rfm_segments.csv",
    "C_monthly":             "module_3_olist/outputs/monthly_customers.csv",
    "C_states":              "module_3_olist/outputs/top_states.csv",
}

out = "powerbi/dashboard_data.xlsx"
missing = []
with pd.ExcelWriter(out) as xw:
    for sheet, path in files.items():
        if not os.path.exists(path):
            missing.append(path); continue
        df = pd.read_csv(path)
        df.columns = [str(c).replace("%", "pct").replace(" ", "_") for c in df.columns]
        df.to_excel(xw, sheet_name=sheet[:31], index=False)
        print(f"✓ {sheet:22} {df.shape}")
print("\nSaved:", out)
if missing: print("\nMISSING (tell me):", *missing, sep="\n  ")
```
